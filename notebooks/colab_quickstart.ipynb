## 1. 본문 작성

마커 뒤에는 공백 한 칸. 번호(Ⅰ. 1.)는 도구가 붙인다.

각주는 근거가 되는 말 **바로 뒤에 붙여** `[^1]`을 적고, 내용은 `[^1]: …` 줄로 적는다(8pt 회색). 문장 전체의 근거이면 마침표 **앞**에 둔다.


````python
%%writefile input.md
# 사업 추진 현황
## 추진 개요
□ 추진 배경 및 목적
○ 제도 개선 요구가 지속 제기되어 개선 방안을 마련[^1]
- 기존 절차의 처리 기간이 길어 이용자 불편이 누적
- 담당 부서별 기준이 달라 안내가 일관되지 않음
○ 이용자 편의 개선과 처리 기간 단축을 목표로 설정
- 목표는 부서 협의를 거쳐 확정
- 세부 지표는 분기별로 점검

| 구분 | 목표 | 실적 |
|---|---|---|
| 처리 기간 | 14일 | 11일 |
| 이용 건수 | 1,000건 | 1,240건 |

:::diagram type=org title="추진 체계"
대표
  기획부
  운영부
  연구부
:::

[^1]: ○○실(2024), 「연도별 추진계획」.
````

## 2. 검사와 변환


````python
!hwpx-studio lint input.md -p policy-default
````

````python
!hwpx-studio build input.md -p policy-default -o 보고서.hwpx --preview preview.html
````

## 5. 도식·조직도 가져오기

웹·문서에 있는 도식(Mermaid·SVG)을 읽어 한글 표로 옮긴다. 색과 점선은 원본 값을 그대로 가져온다.


````python
%%writefile 조직도.mmd
flowchart TD
    A[○○위원회] --> B[기획분과]
    A --> C[운영분과]
    A -.-> D[자문단]
    B --> E[정책팀]
    style A fill:#C00000,color:#FFFFFF
    classDef body fill:#2E75B6,color:#FFFFFF
    class B,C body
````

````python
# 읽은 결과(도식 블록)를 눈으로 확인하고, 그대로 hwpx까지 만든다
!hwpx-studio capture 조직도.mmd --title "위원회 구성" --hwpx 조직도.hwpx
````

원본이 **그림·캡처뿐**이라면 도구가 읽지 못한다. 그때는 아무 AI에게 이렇게 시킨 뒤
나온 블록을 본문(`input.md`)에 붙여 넣으면 된다.

> 이 조직도를 아래 형식으로 옮겨 줘. 2칸 들여쓰기가 한 단계 아래고,
> 상자 색은 `{fill=#RRGGBB color=#RRGGBB}`로 함께 적어 줘.
> ```
> :::diagram type=org title="조직 체계"
> 대표 {fill=#C00000 color=#FFFFFF}
>   기획부 {fill=#2E75B6 color=#FFFFFF}
>   운영부
> :::
> ```

상자가 8개를 넘으면 세로 목록형으로 자동 전환된다(`layout=side`로 직접 지정도 가능).
