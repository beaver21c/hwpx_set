## 1. 본문 작성

마커 뒤에는 공백 한 칸. 번호(Ⅰ. 1.)는 도구가 붙인다.


```python
%%writefile input.md
# 사업 추진 현황
## 추진 개요
□ 추진 배경 및 목적
○ 제도 개선 요구가 지속 제기되어 개선 방안을 마련
- 기존 절차의 처리 기간이 길어 이용자 불편이 누적
- 담당 부서별 기준이 달라 안내가 일관되지 않음
○ 이용자 편의 개선과 처리 기간 단축을 목표로 설정
- 목표는 부서 협의를 거쳐 확정
- 세부 지표는 분기별로 점검

| 구분 | 목표 | 실적 |
|---|---|---|
| 처리 기간 | 14일 | 11일 |
| 이용 건수 | 1,000건 | 1,240건 |

:::diagram type=org title="추진 체계"
대표
  기획부
  운영부
  연구부
:::
```
